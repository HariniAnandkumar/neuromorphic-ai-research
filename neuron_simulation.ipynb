```python
# Simple simulation of a biological neuron membrane

# Membrane potential stores accumulated electrical activity
# Starts at 0 because neuron is initially inactive
membrane_potential = 0.0

# Threshold required for neuron to fire a spike
# If membrane potential reaches this value, neuron emits a spike
threshold = 1.0

# Leak controls how quickly stored charge fades over time
# Biological neurons gradually lose stored voltage if no input arrives
leak = 0.1

# Number of time steps neuron stays inactive after firing
# This simulates the biological refractory period
refractory_steps = 2

# Tracks remaining refractory time
# Starts at 0 because neuron is initially active
refractory_timer = 0

# Input signal arriving at each time step
# Each value represents incoming current to the neuron
inputs = [0.2, 0.3, 0.1, 0.4, 0.2, 0.3, 0.4]

# Loop through all time steps
for t in range(len(inputs)):

    # Check whether neuron is in refractory state
    # During refractory period the neuron cannot fire again
    if refractory_timer > 0:

        print(f"Time {t} | Refractory")

        # Reduce refractory timer each step
        refractory_timer -= 1

        # Skip remaining computations for this step
        continue

    # Apply membrane leakage
    # Stored voltage slowly decays over time
    membrane_potential *= (1 - leak)

    # Add current input into membrane potential
    # Inputs accumulate over time
    membrane_potential += inputs[t]

    # Print current membrane voltage
    print(f"Time {t} | V = {membrane_potential:.2f}")

    # Check whether neuron reaches firing threshold
    if membrane_potential >= threshold:

        # Neuron emits a spike
        print("Spike :)")

        # Reset membrane potential after spike
        # Biological neurons discharge after firing
        membrane_potential = 0.0

        # Activate refractory period
        refractory_timer = refractory_steps
```
